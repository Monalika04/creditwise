# CreditWise AI — Data Understanding

## Objective

Understand, profile and validate the CreditWise application dataset
before performing data cleaning, feature engineering and machine learning.

Dataset:
- 12,000 applications
- 44 columns
- 21 source columns
- 13 synthetic enrichment columns
- 10 derived columns

Target:
- approved

```python
df.info()
```

```python
data_profile = pd.DataFrame({
    "column": df.columns,
    "data_type": df.dtypes.astype(str).values,
    "missing_count": df.isna().sum().values,
    "missing_percentage": (
        df.isna().mean() * 100
    ).round(2).values,
    "unique_values": [
        df[col].nunique(dropna=False)
        for col in df.columns
    ]
})

data_profile
```

```python
missing_report = (
    data_profile[
        data_profile["missing_count"] > 0
    ]
    .sort_values(
        "missing_percentage",
        ascending=False
    )
)

missing_report
```

```python
duplicate_rows = df.duplicated().sum()

print("Duplicate rows:", duplicate_rows)
```

```python
duplicate_ids = df["applicant_id"].duplicated().sum()

print("Duplicate applicant IDs:", duplicate_ids)
```

```python
approval_summary = pd.DataFrame({
    "count": df["approved"].value_counts(),
    "percentage": (
        df["approved"]
        .value_counts(normalize=True)
        .mul(100)
        .round(2)
    )
})

approval_summary
```
