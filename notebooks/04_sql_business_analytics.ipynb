# CreditWise AI — SQL Business Analytics

## Objective

Analyze the CreditWise application portfolio using SQL and identify
patterns related to:

- Approval performance
- Credit score
- Debt-to-income ratio
- Payment history
- Credit utilization
- Income
- Products
- Application channels
- Credit exposure
- Applicant risk

Source:
PostgreSQL warehouse.fact_application

## 1. Business Questions

1. What is the overall approval rate?
2. How many applications are in the portfolio?
3. How does approval vary by product?
4. How does credit score relate to historical approval?
5. How does DTI relate to historical approval?
6. How does payment history relate to approval?
7. How does credit utilization relate to approval?
8. How does income relate to approval?
9. Which applications have high risk indicators?
10. Which applications have the highest requested exposure?
11. How does approval vary by application channel?
12. How does approval vary over time?

```python
SELECT
    COUNT(*) AS total_applications,
    COUNT(DISTINCT applicant_key) AS unique_applicants,
    ROUND(AVG(approved) * 100, 2) AS approval_rate,
    ROUND(AVG(credit_score), 2) AS avg_credit_score,
    ROUND(AVG(debt_to_income_ratio) * 100, 2) AS avg_dti_pct,
    ROUND(AVG(credit_utilization_ratio) * 100, 2) AS avg_utilization_pct,
    ROUND(AVG(requested_credit_limit), 2) AS avg_requested_limit,
    ROUND(SUM(requested_credit_limit), 2) AS total_requested_exposure
FROM warehouse.fact_application;
```

### Result

The portfolio contains 12,000 applications with a historical approval
rate of 39.68%.

Average credit score: 654.07
Average DTI: 22.84%
Average utilization: 22.25%
Average requested limit: approximately ₹60.9K
Total requested exposure: approximately ₹730.72M

These figures describe the synthetic CreditWise demo portfolio.

```python
SELECT
    approved,
    COUNT(*) AS applications,
    ROUND(
        COUNT(*) * 100.0 /
        SUM(COUNT(*)) OVER (),
        2
    ) AS percentage
FROM warehouse.fact_application
GROUP BY approved
ORDER BY approved;
```

### Interpretation

Historical approval distribution:

- Not approved: 60.32%
- Approved: 39.68%

`approved = 0` represents historical non-approval, not default.
